### Crear External Locations requeridas por el proyecto

1. Bronze
2. Silver
3. Gold

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS sadatabricksexternal_bronze
    URL 'abfss://bronze@sadatabricksexternal.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-ext-storage-credential`);
```

```sql
DESC EXTERNAL LOCATION sadatabricksexternal_bronze;
```
