### Crear Catalogs y Schemas para el proyecto

1. Catalog - telco_dev(Sin Managed Location)
2. Schemas - bronze, silver y gold(Con Managed Location)

```sql
CREATE CATALOG IF NOT EXISTS telco_dev;
```

```sql
USE CATALOG telco_dev;
```

```sql
CREATE SCHEMA IF NOT EXISTS bronze
MANAGED LOCATION 'abfss://bronze@sadatabricksexternal.dfs.core.windows.net/bronze';
```

```sql
CREATE SCHEMA IF NOT EXISTS silver
MANAGED LOCATION 'abfss://silver@sadatabricksexternal.dfs.core.windows.net/silver';
```

```sql
CREATE SCHEMA IF NOT EXISTS gold
MANAGED LOCATION 'abfss://gold@sadatabricksexternal.dfs.core.windows.net/gold';
```
